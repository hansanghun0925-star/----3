## 해야 할 일

궁·능 23곳(궁 4 · 종묘 · 조선왕릉 18)의 운영팀에서 아래 일을 의뢰받았다고 가정합니다. 분석 방법은 자유롭게 선택합니다.

1. 다음 며칠의 개소별 관람객 수를 범위로 예측합니다. 예측 기간과 예측 단위는 자유롭게 정합니다
2. 그 범위로 개소별 일일 인력 수(매표·검표·안내·안전을 합친 인원)를 정합니다. 업무별 배분과 언어권별 해설 인력은 선택입니다
3. 인력이 모자랄 때와 남을 때의 비용을 계산해 직전 7일 평균으로 배치했을 때의 비용과 비교합니다

## 지금의 인력 배치 방식

운영팀은 개소마다 인력을 하루 단위로 배치하고, 인원을 **직전 7일 평균**으로 정한다고 가정합니다. 예측 기간 첫날 바로 앞 7일 가운데 문을 연 날의 관람객 평균을 구하고, 이 평균 하나를 예측 기간 모든 날의 예측값으로 씁니다. 이 값을 인력 1명이 하루에 맡는 관람객 수로 나눠 인원을 정합니다. 이 방식은 프로젝트를 위한 가정이므로 보고서에서 사실 근거로 인용하지 않습니다.

이 방식의 오차는 무료 개방일에 몰립니다. 2025년 관람객이 가장 많은 6곳은 경복궁·덕수궁·창덕궁·창경궁·종묘·선릉입니다. 2025년 하반기 이 6곳에서 무료 개방일은 문을 연 날의 13.2%～14.0% 였습니다. 그런데 같은 기간 직전 7일 평균 오차의 32.3%～41.2% 가 무료 개방일에서 나왔습니다.

운영팀의 의뢰입니다.

> "다음 며칠 동안 개소마다 몇 명을 두면 되는지 정해 주세요. 그리고 그 숫자를 왜 믿어야 하는지 알려 주세요."

## 최종 결과물

1. **인력 배치 권고표**: 운영팀이 근무표에 그대로 옮길 수 있는 개소 × 날짜 × 인력 수. 근무표는 데이터 마지막 날(2025-12-31)에 정한다고 가정합니다. 날짜는 2026-01-01부터 팀이 정한 예측 기간이 끝나는 날까지입니다
2. **백테스트 비용**: 같은 방법으로 과거 날짜의 인력을 정했을 때 모자란 인원과 남는 인원의 비용, 직전 7일 평균 대비 절감액
3. **가정을 바꾼 결과**: 서비스 수준이나 비용 가정을 2가지 이상 바꿨을 때 권고 인력 수와 비용이 달라지는 정도

모형끼리 정확도만 비교한 제출물은 요구를 충족하지 못합니다. 인력이 모자랄 때의 손실과 남을 때의 손실은 크기가 다르므로, 어느 쪽 오차를 감수할지 먼저 정합니다. 권고표 날짜(2026년)의 실제 관람객은 아직 알 수 없습니다. 그래서 비용은 백테스트(과거 날짜를 다시 예측해 실제와 비교하는 검증)에서 계산합니다.

## 분석 주제

권고표와 비용 비교는 모든 팀이 냅니다. 가장 자세히 분석할 주제는 자유롭게 선택합니다. 아래 6개 주제에서 골라도 되고, 이 데이터로 답할 수 있는 다른 질문을 정해도 됩니다. 고른 주제와 핵심 질문은 보고서 첫 줄에 적습니다. 질문에는 아래 5단계로 답하고, 효과가 작거나 예측이 나아지지 않았으면 그 결과도 적습니다.

1. 데이터에서 구한 효과의 크기
2. 그 효과를 예측에 반영한 방법
3. 반영 전후의 백테스트 지표
4. 그 차이를 원으로 환산한 비용
5. 운영팀에 권할 규칙(한 문장)

| 주제 | 핵심 질문의 예 | 이 데이터에서 계산한 값 |
|---|---|---|
| 무료 개방일과 명절 연휴 | 무료 개방일에 개소마다 인력을 몇 배로 늘리는가 | 2023～2025년 경복궁의 무료 개방일 80일(문을 연 날의 8.5%)에 관람객의 20.6% 가 왔습니다 |
| 이웃 궁의 휴관 | 경복궁이 쉬는 화요일에 창덕궁은 몇 명을 더 두는가 | 2023～2025년 화요일 창덕궁 관람객은 같은 개소 수～금 평균의 1.51배(외국인만 1.73배)였습니다 |
| 외국인 언어권 | 영어·일어·중국어 해설 인력을 언제 몇 명 두는가 | 궁 4곳 외국인 가운데 영어권 비중은 2019년 29.0% 에서 2025년 53.5% 로 늘고, 중어권은 30.6% 에서 18.3% 로 줄었습니다 |
| 달력에 없는 행사 | 임시 무료 개방이나 행사 공지가 나오면 예측을 어떻게 고치는가 | 종묘대제 날(2025년 5월 4일) 종묘 관람객 44,337명은 평소 같은 요일 평균의 11.61배였습니다 |
| 비 온 날과 그다음 날의 관람객 | 비 예보가 있는 날 인력을 줄이는가 | 10mm 이상 비 온 평일의 궁 관람객은 같은 주 맑은 평일보다 44.4% 적고, 큰비 다음 날이 맑은 주말이면 같은 주 다른 맑은 날보다 43.5% 많았습니다 |
| 직전 7일 평균의 휴관일 처리 | 휴관일을 어떻게 다루느냐에 따라 직전 7일 평균의 지표 값이 얼마나 달라지는가 | 2025년 하반기 경복궁 직전 7일 평균의 WAPE 는 휴관일의 0 을 그대로 두면 50.2%, 문을 연 날만 쓰면 36.7% 입니다 |

WAPE 는 오차 크기의 합을 실제 관람객의 합으로 나눈 값입니다. 표의 무료 개방일은 경복궁이 문을 연 날 가운데 유료 관람객 비중이 15% 미만인 날입니다. 이 정의는 지난 기록을 분석할 때만 씁니다. 예측에 쓰는 무료 개방일 표시는 달력 규칙과 미리 나온 공지로 만듭니다(「자주 하는 실수」 3번). 팀이 다른 정의를 쓰면 보고서에 적습니다.

<details class="lms-extra"><summary>표의 값을 계산한 방법</summary>

- **평소**: 2023～2025년에 문을 연 날 가운데 무료 개방일 · 공휴일 · 명절 연휴(설날 · 추석 당일을 포함해 이어지는 휴일)가 아닌 날입니다. 표의 몇 배 값(11.61배 · 1.51배)과 비 효과는 이런 날만으로 계산했습니다
- **종묘대제 11.61배**: 그날 종묘 관람객을 앞뒤 28일 안의 평소 같은 요일 평균으로 나눴습니다
- **창덕궁 화요일 1.51배**: 평소의 화요일 평균을 같은 개소의 평소 수～금 평균으로 나눴습니다
- **비 온 평일 44.4%**: 같은 개소 · 주 · 날짜 유형 안에서 비 온 날 평균을 맑은 날 평균으로 나눈 비율의 로그 중앙값입니다. 맑은 날은 `precip_mm` 1mm 미만입니다. 비 온 날의 기준을 1mm 이상으로 낮추면 비 온 평일 관람객은 맑은 평일보다 16.5% 적습니다
- **큰비 다음 날 43.5%**: 10mm 이상 비 온 다음 날이 맑은 경우를 같은 주에서 전날도 맑았던 다른 맑은 날과 비교했습니다
- **경복궁 WAPE 50.2% · 36.7%**: 2025-07-01부터 7일 간격의 예측 원점 26개에서 원점마다 7일을 예측했습니다. 36.7% 는 문을 연 날만 평균하고 평가한 값입니다
- **외국인 언어권 비중**: 궁 4곳 외국인 4개 열의 합에서 각 언어권이 차지하는 비율입니다. 종묘는 기록이 2023년부터라 뺐습니다

## 선택 과제

요구사항 8가지를 모두 채운 뒤 해 볼 수 있는 과제입니다. 결과는 보고서의 「추가 발견과 다음 할 일」에 적습니다.

- **업무별 인력 배분**: 개소별 인력 수를 매표·검표·안내·안전으로 나눕니다. 무료 개방일에는 매표가 줄고 안내·안전이 늘어납니다
- **관리소 단위 인력 공유**: 같은 관리소의 왕릉끼리 인력을 함께 쓰면 관람객이 적은 왕릉의 최소 인원 문제가 줄어드는지 확인합니다

### 데이터 주의 사항

값은 원자료 그대로이고, 고치거나 채운 값은 없습니다. 분석 전에 아래를 직접 확인하고, 찾은 내용과 처리한 방법을 보고서의 「오차 분석과 한계」에 적습니다.

- **휴관일과 0**: 정기 휴관 요일(`sites.csv` 의 `closed_weekday`)과 관람객이 0 인 날이 늘 일치하는지
- **빠진 날짜**: 날짜가 통째로 빠진 기간이 있는지
- **열의 합**: 「유료 + 무료」와 「내국인 + 외국인」이 맞지 않는 행이 있는지
- **이상값**: 음수나 빈 값이 있는지
- **언어권 열**: 유료 관람객이 0 인 날 언어권 분포가 어떻게 달라지는지. 언어권을 나눈 기준은 원본에 설명이 없습니다
- **개소별 기간과 수준**: 개소마다 자료가 시작하는 해와 코로나 전후 관람객 수준이 어떻게 다른지

## 공변량 출처

공변량의 출처는 아래 3가지로 한정합니다.

- **날씨**: 데이터 폴더의 `weather_daily.parquet` 입니다. NASA POWER 일별 API 에서 개소 좌표마다 받은 값이고 키는 `date`·`site` 입니다. 기온 `t2m` · `t2m_max` · `t2m_min` 의 단위는 ℃, 강수량 `precip_mm` 의 단위는 mm/일입니다. 비 온 날의 강수량 기준은 자유롭게 정하고 보고서에 적습니다
- **달력**: 데이터 폴더의 `calendar_kr_2015_2025.csv` 입니다. `holidays` 패키지(0.103)의 `holidays.KR` 로 만든 공휴일, 설날 · 추석까지의 날 수, 매월 마지막 수요일 열이 있습니다. 2026년 권고표 날짜의 달력은 같은 패키지로 만듭니다. 다른 나라 공휴일은 `holidays.CN` · `holidays.JP` 로 만들 수 있고, 직접 정한 날짜 규칙(예: 일본 골든위크 4월 29일～5월 5일)도 쓸 수 있습니다
- **행사·임시 무료 개방 날짜**: 국가유산청·궁능유적본부의 공식 공지에서 확인하고 그 주소를 보고서에 적습니다. 「문화가 있는 날」이 매주 수요일로 바뀌어 2026년 5월 27일 조선왕릉부터 매주 수요일 무료 개방이 시작됐습니다. 덕수궁 · 창덕궁 · 창경궁 · 종묘에도 차례로 적용한다고 발표했습니다. 권고표 날짜(2026년 1월부터)에는 해당하지 않으므로 보고서의 「추가 발견과 다음 할 일」에서 다룹니다

날씨 파일은 NASA POWER 의 격자 자료라 가까운 개소는 같은 값을 받고, 국지적인 비는 반영되지 않습니다. 기온과 강수량을 그대로 쓸지, 비 온 날 여부 같은 열로 바꿔 쓸지는 직접 확인하고 정합니다.

### 날씨 값 사용 규칙

공변량은 예측하는 시점에 알 수 있는 값만 씁니다. NASA POWER 의 값은 그날의 실제 날씨를 재분석한 값이라 근무표를 정하는 날에는 알 수 없습니다. 날씨 값은 아래 2가지 가운데 하나로 씁니다.

1. **지난 날의 실측**: 근무표를 정할 때 이미 지난 날의 실측만 씁니다. 기본은 원점 전날까지입니다
2. **완벽한 예보 가정**: 예측할 날의 실측(당일 실측)을 씁니다. 예보가 실측과 똑같이 맞았다고 가정한 결과로, 날씨 정보로 얻을 수 있는 개선의 상한입니다. 비교표에 「완벽한 예보 가정」으로 표시하고 1번이나 날씨 정보가 없는 결과와 나란히 적습니다

당일 실측을 표시 없이 다른 결과와 함께 쓰면 미래 정보 누수로 봅니다. 날씨 파일은 2025-12-31 까지라 권고표 날짜에는 1번 방식(지난 날의 실측)만 쓸 수 있습니다.

## 요구사항

제출물에 아래 8가지가 모두 있어야 합니다. 방법은 자유롭게 선택합니다.

**1. 예측 단위 · 예측 기간 · 1순위 지표**

예측 단위, 예측 기간, 1순위 평가 지표를 정합니다. 예측 단위는 개소별 · 관리소별 · 개소 유형별 합계 가운데 하나이고, 예측 기간은 근무표가 다루는 날 수입니다. 근무표를 언제 확정하는지 가정하고, 확정하는 날까지의 데이터만 씁니다. 지표 값은 인력을 배치하는 단위(개소 × 날짜)로 계산하고, 휴관일은 평가에서 뺀 뒤 그 규칙을 적습니다. 하루 관람객이 적은 개소에서는 MAPE 가 불안정해지므로 지표는 개소 크기를 고려해 고릅니다.

> **보고서**: 「질문과 범위」에 예측 단위 · 예측 기간 · 1순위 지표를 적습니다.

**2. 시간 순서를 지킨 백테스트와 베이스라인**

예측하는 기간의 첫날을 **예측 원점(forecast origin)** 이라고 합니다. 예측 원점은 5개 이상 정하고, 원점마다 근무표 확정 시점(기본은 원점 전날)까지의 데이터로만 학습 · 전처리 · 예측합니다. 지표 값은 원점별로도 보고합니다.

**베이스라인(baseline)** 은 과거 값으로 예측값을 정하는 단순한 규칙입니다. 아래 3개를 최종 모형과 같은 원점 · 같은 지표로 계산해 비교표에 함께 적습니다. 비교표는 「전체」와 「무료 개방일·공휴일을 뺀 날」로 나눠 내기를 권합니다.

- **나이브 기법**: 원점 전 마지막 개방일의 값
- **계절 나이브(7일)**: 원점 전 7일 가운데 같은 요일의 값. 그날이 휴관일이면 문을 연 가장 가까운 같은 요일의 값
- **직전 7일 평균**: 「지금의 인력 배치 방식」의 규칙

> **보고서**: 「근거와 재현 조건」에 예측 원점과 원점별 학습 구간을 날짜로 적습니다. 비교표에는 베이스라인 3개와 최종 모형의 원점별 지표 값이 있어야 합니다.

**3. 범위 예측과 인력 수**

관람객을 예측 구간이나 분위수로 예측하고, 그 범위의 어느 값까지 대비할지(서비스 수준) 정해 인력 수로 바꿉니다. 서비스 수준이 0.8(기본 가정)이고 인력 1명이 하루 2,000명을 맡으면, 관람객 0.8 분위수 ÷ 2,000 명을 배치합니다. 필요 인력은 실제 관람객 ÷ 2,000 명입니다. 구간으로 냈다면 백테스트에서 구간이 실제 관람객을 포함한 비율(**포함률**)을 확인합니다. 분위수 하나로 냈다면 필요 인력이 배치 인력 이하였던 비율(**적중률**)을 확인합니다. 적중률 계산과 아래 기본 가정 · 임계비율 식은 딥러닝 시계열 과목에서 서울 지하철 역 인력을 정할 때와 같습니다.

> **보고서**: 「선택과 이유」에 인력 수가 관람객 분포의 어느 분위수인지 적습니다. 「근거와 재현 조건」에 포함률(구간을 쓴 경우)이나 적중률(분위수 하나로 정한 경우)을 숫자로 적습니다.

**4. 비용 환산과 절감액**

팀끼리 결과를 비교할 수 있도록 아래 **기본 가정**을 씁니다.

- 인력이 하루 1명 모자라면 60만 원, 하루 1명 남으면 15만 원의 비용으로 봅니다. 이 두 금액을 각각 $C_u$, $C_o$ 라고 부릅니다
- 인력 1명은 하루 관람객 2,000명을 맡습니다
- 하루 비용은 $C_u$ × 모자란 명수 + $C_o$ × 남는 명수입니다. 인력 수는 올림하지 않고 소수 그대로 계산합니다

2025-10-14(화) 창덕궁 인력을 직전 7일 평균으로 정하면 아래와 같습니다.

- 직전 7일(10-07～10-13) 가운데 문을 연 6일의 평균은 25,833명입니다. 이 6일에는 추석 연휴 무료 개방일 3일(10-07～10-09)이 있습니다
- 배치 인력은 25,833 ÷ 2,000 = 12.9명, 필요 인력은 실제 관람객 7,297 ÷ 2,000 = 3.6명입니다
- 남는 9.3명의 비용은 9.3 × 15만 원 = 139만 5천 원입니다. 반올림하지 않은 9.2678명이면 139만 175원입니다
- 무료 개방일 3일의 평균은 45,886명, 나머지 3일은 5,779명이라 이 3일이 평균을 크게 높였습니다

백테스트의 모든 원점 · 개소 · 날의 비용을 더해 직전 7일 평균의 비용과 비교하면 절감액이 나옵니다. 절감액은 둘로 나눕니다. 직전 7일 평균을 0.8 분위수에 맞추기만 해서(서비스 수준 보정, 요구사항 6의 2행) 줄어든 금액과, 모형을 써서 더 줄어든 금액입니다. 가정을 바꾸면 이유와 기본 가정의 결과를 함께 적습니다. 팀이 더한 규칙(올림 · 최소 인원 등)은 직전 7일 평균에도 똑같이 적용합니다.

> **보고서**: 「근거와 재현 조건」에 모자란 인원과 남는 인원의 비용, 직전 7일 평균의 비용, 기본 가정에서의 절감액을 원 단위로 적습니다. 절감액은 서비스 수준 보정만으로 줄어든 금액과 모형으로 더 줄어든 금액으로 나눕니다. 「오차 분석과 한계」에 가정을 바꿔 다시 계산한 결과를 2가지 이상 적습니다.

**5. 재현 조건**

다른 사람이 보고서만 보고 같은 숫자를 다시 만들 수 있어야 합니다. 채점자는 코드를 실행하지 않고 보고서의 재현 조건을 봅니다.

> **보고서**: 「근거와 재현 조건」 끝에 사용한 데이터 구간 · 분할 방식 · 시드 · 주요 설정과, 검산값 7개가 맞았다는 문장을 적습니다.

**6. FVA 분석: 정보를 하나씩 더할 때의 비용 변화**

수요예측에서는 예측 과정의 단계마다 지표와 비용이 얼마나 나아졌는지 측정해, 효과가 없는 단계를 뺍니다. 이 방법을 FVA(Forecast Value Added, 예측 부가가치) 분석이라고 합니다. 결과는 표 하나로 냅니다. 행의 순서는 아래와 같고, 행마다 더한 정보 · 1순위 지표 · 비용(원) · 앞 행보다 줄어든 비용을 적습니다. 비용이 늘어난 정보도 남깁니다.

1. **1행**: 운영팀이 지금 쓰는 직전 7일 평균
2. **2행**: 직전 7일 평균에 서비스 수준 보정만 더한 결과. 예를 들어 원점 전 몇 주의 「실제 관람객 ÷ 직전 7일 평균」에서 구한 0.8 분위수를 곱합니다
3. **3행부터**: 팀 모형으로 계산합니다. 3행은 과거 관람객만 씁니다. 그다음 행마다 팀이 고른 정보를 하나씩 더합니다. 요일 · 정기 휴관 요일, 공휴일 · 명절 연휴, 매월 마지막 수요일, 이웃 개소의 휴관, 날씨 등입니다. 당일 실측 날씨는 「완벽한 예보 가정」 행에 따로 적습니다

이 표는 후보 모형으로 먼저 만들고, 최종 모형을 정한 뒤 다시 계산합니다.

> **보고서**: 「근거와 재현 조건」에 FVA 표를 적습니다.

**7. 오차가 큰 날의 분석과 보정 규칙**

백테스트 결과를 아래 유형으로 나눠 1순위 지표와 적중률(구간을 쓴 팀은 포함률)을 계산합니다.

- 날짜 유형: 평일·주말, 공휴일, 무료 개방일
- 개소 유형: 궁, 종묘, 왕릉

### 서비스 수준

서비스 수준은 인력 수를 관람객 분포의 어느 분위수에 맞출지 나타내는 값이며, **임계비율(critical ratio)** 로 정합니다. 식은 $\tau^* = C_u / (C_u + C_o)$ 입니다. 기본 가정에서는 $60 \div (60 + 15) = 0.8$ 이라 0.8 분위수에 맞춥니다. 무료 개방일처럼 모자랄 때의 손실이 큰 날에 $C_u$ 를 크게 둘 수도 있지만, 비용 데이터가 없으므로 그 금액은 팀의 가정으로 적습니다.

## 자주 하는 실수

아래 5가지는 이 데이터에서 실제로 일어나며 결과만 보고는 알아차리기 어렵습니다. 팀 분석에서 확인한 것은 보고서의 「오차 분석과 한계」에 적습니다. 그 실수가 결과를 얼마나 바꾸는지 숫자로 측정하고(예: 휴관일을 빼기 전후의 직전 7일 평균 WAPE 차이), 어떻게 막았는지 씁니다.

1. **휴관일의 0 을 수요로 학습**: 휴관일의 0 을 관람 수요 0 으로 학습하면 정기 휴관 요일에 문을 연 날에도 관람객이 없다고 예측합니다
2. **빠진 날짜 구간**: 빠진 날짜를 0 으로 채우면 1번과 같은 문제가 생깁니다. 행 순서대로 지난 값을 가져오면 7일 전 값 대신 훨씬 이전의 값을 쓰게 됩니다
3. **미래 정보 누수**: 날씨 당일 실측, `paid == 0` 으로 만든 무료 개방일 표시, 전체 데이터로 먼저 만든 표준화·인코딩은 예측 시점에 알 수 없는 값입니다. 무료 개방일은 달력 규칙(명절 연휴·매월 마지막 수요일)과 배치 전에 나온 공지로 정합니다
4. **합계로 평가**: 합계로 지표 값을 계산하면 개소 간 +오차와 −오차가 상쇄되어 실제보다 좋게 나옵니다. 권고표는 개소마다 인원을 정하므로 지표 값도 개소 단위로 계산합니다
5. **유료 0 을 무료 개방일로 봄**: 개소마다 유료 기록 방식이 달라, 무료 개방일에도 유료 관람객이 기록되는 개소가 있습니다. 그래서 무료 개방일은 개소별 `paid` 값이 아니라 달력 규칙과 공지로 정합니다
